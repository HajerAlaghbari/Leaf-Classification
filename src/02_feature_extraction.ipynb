```python
import os
import cv2
import numpy as np
import pandas as pd
from skimage.feature import greycomatrix, greycoprops
```

```python
import sys
print(sys.executable)
```

```python
DATA_PATH = "../data/processed/"
FEATURES_PATH = "../data/features/leaf_features.csv"
```

```python
# --- 1. Function to extract GLCM + Template features ---
def extract_features(img):
    # GLCM features
    glcm = greycomatrix(img, [1], [0], symmetric=True, normed=True)
    contrast = greycoprops(glcm, 'contrast')[0,0]
    energy = greycoprops(glcm, 'energy')[0,0]
    homogeneity = greycoprops(glcm, 'homogeneity')[0,0]
    correlation = greycoprops(glcm, 'correlation')[0,0]
    
    # Template Matching score placeholder (mean pixel intensity)
    template_score = np.mean(img)
    
    return [contrast, energy, homogeneity, correlation, template_score]
```

```python
# --- 2. Iterate over images and extract features ---
features_list = []
labels_list = []

for species in os.listdir(DATA_PATH):
    species_folder = os.path.join(DATA_PATH, species)
    for img_name in os.listdir(species_folder):
        img_path = os.path.join(species_folder, img_name)
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        img = cv2.resize(img, (128,128))
        
        features = extract_features(img)
        features_list.append(features)
        labels_list.append(species)
```

```python
# --- 3. Save features to CSV ---
features_df = pd.DataFrame(features_list, columns=['contrast','energy','homogeneity','correlation','template_score'])
features_df['label'] = labels_list
features_df.to_csv(FEATURES_PATH, index=False)
print("Features extracted and saved to CSV!")
```
