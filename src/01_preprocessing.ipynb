Leaf Classification - Preprocessing

```python
import os
import zipfile
import pandas as pd
import cv2
import shutil
```

```python
# --- 1. Paths ---
RAW_PATH = "../data/raw/"
PROCESSED_PATH = "../data/processed/"
IMAGES_ZIP = os.path.join(RAW_PATH, "images.zip")
TRAIN_CSV = os.path.join(RAW_PATH, "train.csv.zip")
```

```python
# --- 2. Unzip images ---
with zipfile.ZipFile(IMAGES_ZIP, 'r') as zip_ref:
    zip_ref.extractall(os.path.join(RAW_PATH, "images"))
```

```python
# --- 3. Read train.csv ---
train_df = pd.read_csv(TRAIN_CSV, compression='zip')
print(train_df.head())
```

```python

# --- 4. Organize images into folders by species ---
if not os.path.exists(PROCESSED_PATH):
    os.makedirs(PROCESSED_PATH)

for idx, row in train_df.iterrows():
    img_id = str(row['id']) + ".jpg"
    species = row['species']
    
    species_folder = os.path.join(PROCESSED_PATH, species)
    if not os.path.exists(species_folder):
        os.makedirs(species_folder)
    
    src_img = os.path.join(RAW_PATH, "images", img_id)
    dst_img = os.path.join(species_folder, img_id)
    
    if os.path.exists(src_img):
        shutil.copy(src_img, dst_img)

print("Images organized into folders by species!")
```
